````python
import pandas as pd

print("Pandas version:", pd.__version__)
print("Libraries imported successfully!")
````

````python
data = {
    "Employee_ID": [105, 102, 104, 101, 103, 106, 107, 108],
    "Name": [
        "Arun", "Divya", "Karthik", "Priya",
        "Rahul", "Meena", "Vijay", "Anu"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "HR"
    ],
    "Age": [25, 30, 28, 24, 32, 27, 26, 29],
    "Salary": [
        35000, 45000, 40000, 30000,
        50000, 38000, 42000, 46000
    ]
}

df = pd.DataFrame(data)

print("Employee Dataset Created Successfully!")
df
````

````python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
````

````python
ascending_salary = df.sort_values(
    by="Salary",
    ascending=True
)

print("Employees Sorted by Salary: Lowest to Highest")
display(ascending_salary)
````

````python
descending_salary = df.sort_values(
    by="Salary",
    ascending=False
)

print("Employees Sorted by Salary: Highest to Lowest")
display(descending_salary)
````

````python
ascending_name = df.sort_values(
    by="Name",
    ascending=True
)

print("Employees Sorted by Name: A to Z")
display(ascending_name)
````

````python
descending_name = df.sort_values(
    by="Name",
    ascending=False
)

print("Employees Sorted by Name: Z to A")
display(descending_name)
````

````python
sorted_department = df.sort_values(
    by="Department",
    ascending=True
)

print("Employees Sorted by Department: A to Z")
display(sorted_department)
````

````python
ascending_age = df.sort_values(
    by="Age",
    ascending=True
)

print("Employees Sorted by Age: Youngest to Oldest")
display(ascending_age)
````

````python
descending_age = df.sort_values(
    by="Age",
    ascending=False
)

print("Employees Sorted by Age: Oldest to Youngest")
display(descending_age)
````

````python
multi_column_sort = df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)

print("Sorted by Department (A-Z) and Salary (High to Low)")
display(multi_column_sort)
````

````python
age_salary_sort = df.sort_values(
    by=["Age", "Salary"],
    ascending=[True, True]
)

print("Sorted by Age and Salary in Ascending Order")
display(age_salary_sort)
````

````python
sorted_employee_id = df.sort_values(
    by="Employee_ID",
    ascending=True
)

print("Employees Sorted by Employee ID")
display(sorted_employee_id)
````

````python
reset_sorted_df = df.sort_values(
    by="Salary",
    ascending=False
).reset_index(drop=True)

print("Sorted Dataset with Reset Index")
display(reset_sorted_df)
````

````python
print("Original Dataset:")
display(df)

temporary_sort = df.sort_values(
    by="Salary",
    ascending=False
)

print("Sorted Dataset:")
display(temporary_sort)

print("Original Dataset Remains Unchanged:")
display(df)
````

````python
df_sorted = df.sort_values(
    by="Salary",
    ascending=False
).copy()

print("Dataset Sorted and Stored in a New DataFrame:")
display(df_sorted)
````

````python
print("Ascending Salary Order:")
display(df.sort_values(by="Salary", ascending=True)[["Name", "Salary"]])

print("Descending Salary Order:")
display(df.sort_values(by="Salary", ascending=False)[["Name", "Salary"]])
````

````python
top_3_employees = df.sort_values(
    by="Salary",
    ascending=False
).head(3)

print("Top 3 Highest-Paid Employees:")
display(top_3_employees[["Name", "Department", "Salary"]])
````

````python
print("========== TASK 20 SUMMARY ==========")

print("\n1. Original Dataset:")
display(df)

print("\n2. Ascending Salary:")
display(df.sort_values(by="Salary", ascending=True))

print("\n3. Descending Salary:")
display(df.sort_values(by="Salary", ascending=False))

print("\n4. Categorical Sorting:")
display(df.sort_values(by="Department", ascending=True))

print("\n5. Multiple Column Sorting:")
display(
    df.sort_values(
        by=["Department", "Salary"],
        ascending=[True, False]
    )
)

print("\nTASK 20 COMPLETED SUCCESSFULLY!")
````

### 1. How do you sort a DataFrame?

We use the Pandas `sort_values()` method to sort a DataFrame by one or more columns.

Example:

```python
df.sort_values(by="Salary")
```

### 2. What does `ascending=False` do?

It sorts values in descending order. For numerical values, the largest value appears first.

Example:

```python
df.sort_values(by="Salary", ascending=False)
```

### 3. Can a DataFrame be sorted by multiple columns?

Yes. We can provide multiple column names in a list and specify the sorting direction for each column.

Example:

```python
df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)
```

### 4. Does `sort_values()` modify the original DataFrame?

By default, it returns a sorted DataFrame without modifying the original DataFrame. We can assign the result to a variable or use `inplace=True` to modify the existing DataFrame.

### 5. What is the difference between ascending and descending order?

* **Ascending:** Smallest to largest, or A to Z.
* **Descending:** Largest to smallest, or Z to A.
